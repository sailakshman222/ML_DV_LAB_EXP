# Experiment 6: Pie and Donut

**Aim:** Analyze sales proportions by category.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default='notebook_connected'
DATA=Path('data') if Path('data').exists() else Path('DV/data')
df=pd.read_csv(DATA/'retail_sales.csv',parse_dates=['Date'])

a=df.groupby('Category',as_index=False).Sales.sum()
a['SharePercent']=100*a.Sales/a.Sales.sum()
print(a.to_string(index=False))
px.pie(a,names='Category',values='Sales',title='Sales share: pie').show()
px.pie(a,names='Category',values='Sales',hole=.45,title='Sales share: donut').show()
```
