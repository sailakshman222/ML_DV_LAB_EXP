# Experiment 7: Heatmaps and Treemaps

**Aim:** Show correlations and multilevel category sales.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default='notebook_connected'
DATA=Path('data') if Path('data').exists() else Path('DV/data')
df=pd.read_csv(DATA/'retail_sales.csv',parse_dates=['Date'])

corr=df[['Quantity','Sales','Profit']].corr()
print(corr.to_string())
px.imshow(corr,text_auto='.2f',zmin=-1,zmax=1,color_continuous_scale='RdBu_r',title='Correlation heatmap').show()
a=df.groupby(['Category','Subcategory'],as_index=False).Sales.sum()
px.treemap(a,path=['Category','Subcategory'],values='Sales',title='Category and subcategory sales').show()
```
