# Experiment 1: Getting Started

**Aim:** Import CSV and create a table, bar chart and line chart.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default='notebook_connected'
DATA=Path('data') if Path('data').exists() else Path('DV/data')
df=pd.read_csv(DATA/'retail_sales.csv',parse_dates=['Date'])

print(df.head().to_string(index=False))
a=df.groupby('Category',as_index=False).Sales.sum()
px.bar(a,x='Category',y='Sales',title='Category sales').show()
b=df.set_index('Date').Sales.resample('MS').sum().reset_index()
px.line(b,x='Date',y='Sales',markers=True,title='Monthly sales').show()
```
