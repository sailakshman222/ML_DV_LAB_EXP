# Experiment 5: Geo Visualization

**Aim:** Create a city-wise sales symbol map.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default='notebook_connected'
DATA=Path('data') if Path('data').exists() else Path('DV/data')
df=pd.read_csv(DATA/'retail_sales.csv',parse_dates=['Date'])

a=df.groupby(['City','State','Latitude','Longitude'],as_index=False).Sales.sum()
print(a.to_string(index=False))
fig=px.scatter_geo(a,lat='Latitude',lon='Longitude',size='Sales',color='State',hover_name='City',title='City-wise sales in India')
fig.update_geos(fitbounds='locations',showland=True,showcountries=True)
fig.show()
```
