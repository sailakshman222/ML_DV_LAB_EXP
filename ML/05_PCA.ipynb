# Experiment 5: PCA

**Aim:** Reduce Iris features from four to two.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
iris=load_iris(); X=StandardScaler().fit_transform(iris.data)
pca=PCA(n_components=2); Z=pca.fit_transform(X)
print('Original:',X.shape,'Reduced:',Z.shape)
print('Explained variance:',pca.explained_variance_ratio_)
for i,name in enumerate(iris.target_names):
    plt.scatter(Z[iris.target==i,0],Z[iris.target==i,1],label=name)
plt.xlabel('PC1'); plt.ylabel('PC2'); plt.legend(); plt.show()
```
