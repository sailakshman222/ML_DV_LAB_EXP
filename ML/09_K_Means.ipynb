# Experiment 9: K Means

**Aim:** Cluster unlabeled Iris measurements.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
X=StandardScaler().fit_transform(load_iris().data)
ks=range(1,11)
plt.plot(list(ks),[KMeans(n_clusters=k,n_init=10,random_state=42).fit(X).inertia_ for k in ks],'o-')
plt.xlabel('K'); plt.ylabel('Inertia'); plt.show()
model=KMeans(n_clusters=3,n_init=10,random_state=42); labels=model.fit_predict(X)
print('Silhouette score:',silhouette_score(X,labels))
pca=PCA(n_components=2).fit(X); Z=pca.transform(X); centers=pca.transform(model.cluster_centers_)
plt.scatter(Z[:,0],Z[:,1],c=labels,cmap='viridis'); plt.scatter(centers[:,0],centers[:,1],marker='X',s=150,c='red')
plt.xlabel('PC1'); plt.ylabel('PC2'); plt.show()
```
