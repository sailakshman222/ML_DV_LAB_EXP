# Experiment 1: Simple Linear Regression

**Aim:** Predict marks from study hours.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,r2_score
X=np.arange(1,13).reshape(-1,1)
y=np.array([32,37,41,49,51,57,62,68,73,79,84,90])
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.25,random_state=42)
model=LinearRegression().fit(X_train,y_train)
pred=model.predict(X_test)
print('Slope:',model.coef_[0],'Intercept:',model.intercept_)
print('MAE:',mean_absolute_error(y_test,pred),'R2:',r2_score(y_test,pred))
plt.scatter(X.ravel(),y); plt.plot(X.ravel(),model.predict(X),'r-')
plt.xlabel('Hours'); plt.ylabel('Marks'); plt.show()
```
